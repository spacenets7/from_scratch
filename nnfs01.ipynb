```python
inputs = [1,2,3,2.5]

weights = [[0.2, 0.8,-0.5,1],
          [0.5, -0.91,0.26,-0.5],
          [-0.26, -0.27,0.17,0.87]]

biases = [2,3,0.5]

layer_outputs = []  # output of current layer
for neuron_weights, neuron_bias in zip(weights, biases): #zip converts two lists into one element-wise
    neuron_output = 0 #output of given neuron
    for n_input, weight in zip(inputs, neuron_weights):
        neuron_output += neuron_bias
        layer_outputs.append(neuron_output)

print(layer_outputs)
```

```python
import numpy as np
# tensor is an object that can be represented as an array in deep learning
inputs = [1,2,3,2.5]
weights = [0.2, 0.8,-0.5,1]
bias = 2

output = np.dot(inputs, weights) + bias
print(output)
```
